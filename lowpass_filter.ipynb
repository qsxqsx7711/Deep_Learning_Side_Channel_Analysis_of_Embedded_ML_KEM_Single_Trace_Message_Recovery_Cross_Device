```python
import numpy as np
import scipy.signal as signal
import matplotlib.pyplot as plt

# Sampling frequency
fs = 1000
# Normalized cutoff for design (Nyquist = fs/2)
ws = 100 / (fs / 2)

# List of filter orders to compare
orders = [1, 5, 10, 15, 20, 25, 30]
# Cutoff frequency in Hz
fc = 100

plt.figure(figsize=(8, 6))

# Plot frequency responses using SOS
for order in orders:
    sos = signal.butter(order, ws, btype='lowpass', output='sos')
    w, h = signal.sosfreqz(sos, worN=4096)
    freq = w * fs / (2 * np.pi)
    plt.plot(freq, np.abs(h), label=f'Order {order}')

# Mark cutoff frequency
plt.axvline(fc, color='k', linestyle='--', label='Cutoff (100 Hz)')

# Limit x-axis to 500 Hz
plt.xlim(0, 500)
plt.ylim(0, 1.05)

plt.title('Butterworth Lowpass Filter Frequency Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.legend(ncol=2, fontsize='small')
plt.grid(True)
plt.tight_layout()
plt.savefig(f'figure/lowpass_filter/butterworth.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import scipy.signal as signal
import matplotlib.pyplot as plt

# Sampling frequency
fs = 1000
# Normalized cutoff for design (Nyquist = fs/2)
ws = 100 / (fs / 2)

# Passband ripple for Chebyshev Type I
rp = 1

# List of filter orders to compare
orders = [1, 5, 10, 15, 20, 25, 30]
# Cutoff frequency in Hz
fc = 100

plt.figure(figsize=(8, 6))

# Plot frequency responses using SOS for Chebyshev Type I
for order in orders:
    sos = signal.cheby1(order, rp, ws, btype='lowpass', output='sos')
    w, h = signal.sosfreqz(sos, worN=4096)
    freq = w * fs / (2 * np.pi)
    plt.plot(freq, np.abs(h), label=f'Order {order}')

# Mark cutoff frequency
plt.axvline(fc, color='k', linestyle='--', label='Cutoff (100 Hz)')

# Limit x-axis to 500 Hz
plt.xlim(0, 500)
plt.ylim(0, 1.05)

plt.title('Chebyshev Type I Lowpass Filter Frequency Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.legend(ncol=2, fontsize='small')
plt.grid(True)
plt.tight_layout()
plt.savefig(f'figure/lowpass_filter/chebyshev.png', dpi=200, bbox_inches='tight')
plt.show()
```
