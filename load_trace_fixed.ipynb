```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
trace_num = 2000

trace_path = f'traces/traces_fixed_message_{trace_num}'
trace_path_new = f'traces/traces_fixed_message_{trace_num}_new'

t = np.load(f'{trace_path}/trace.npy')
m = np.load(f'{trace_path}/message.npy')

t_new = np.load(f'{trace_path_new}/trace.npy')
m_new = np.load(f'{trace_path_new}/message.npy')
```

```python
print(t.shape)
print(m.shape)
print(m)

print(t_new.shape)
print(m_new.shape)
print(m_new)
```

```python
print(np.max(t[0]))
print(np.max(t_new[0]))
```

```python
plt.figure(figsize=(16, 8))
plt.rc('xtick', labelsize=20)   # x 軸刻度字體大小
plt.rc('ytick', labelsize=20)   # y 軸刻度字體大小
plt.xlabel('Sample Points', fontsize=20)
plt.ylabel('Power Consumption', fontsize=20)
plt.plot(t[0])
plt.savefig('figure/load_trace/unfixed_message_2000_single_trace.png', dpi=200, bbox_inches='tight')
```

```python
plt.figure(figsize=(16, 8))
plt.rc('xtick', labelsize=20)   # x 軸刻度字體大小
plt.rc('ytick', labelsize=20)   # y 軸刻度字體大小
plt.xlabel('Sample Points', fontsize=20)
plt.ylabel('Power Consumption', fontsize=20)
plt.plot(t_new[0])
plt.savefig('figure/load_trace/unfixed_message_2000_single_trace.png', dpi=200, bbox_inches='tight')
```

```python
plt.figure(figsize=(16, 8))
plt.rc('xtick', labelsize=20)   # x 軸刻度字體大小
plt.rc('ytick', labelsize=20)   # y 軸刻度字體大小
plt.xlabel('Sample Points', fontsize=20)
plt.ylabel('Power Consumption', fontsize=20)
plt.plot(np.mean(t, axis=0))
plt.savefig('figure/load_trace/unfixed_message_2000_mean_trace.png', dpi=200, bbox_inches='tight')
```

```python
plt.figure(figsize=(16, 8))
plt.rc('xtick', labelsize=20)   # x 軸刻度字體大小
plt.rc('ytick', labelsize=20)   # y 軸刻度字體大小
plt.xlabel('Sample Points', fontsize=20)
plt.ylabel('Power Consumption', fontsize=20)
plt.plot(np.mean(t_new, axis=0))
plt.savefig('figure/load_trace/unfixed_message_2000_mean_trace.png', dpi=200, bbox_inches='tight')
```
