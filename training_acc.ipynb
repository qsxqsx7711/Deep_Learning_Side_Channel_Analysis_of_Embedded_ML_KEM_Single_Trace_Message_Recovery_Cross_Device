```python
import pandas as pd

# 1. 請把路徑改成你自己 CSV 檔案的實際路徑
csv_path = "figure/train_mlp_best_dataset/CW308_STM32F4/1000/results/val_accuracy_comparison.csv"

# 2. 讀取 CSV
df = pd.read_csv(csv_path)

# 3. 假設第一欄是 epoch，後面都是不同 dataset 對應的 accuracy
#    我們先把 epoch 欄取出來，方便之後回頭查是哪一個 epoch
epochs = df["epoch"]

# 4. 把所有 Accuracy 欄位都挑出來（這裡用 .iloc 來定位：第一欄之後都是 accuracy）
#    也可以用 df.filter(regex="000Traces_Pol") 之類的方式，只要能選到正確的欄位都行
acc_df = df.iloc[:, 1:]  # 從第二欄開始（index=1）到最後一欄，都當作 accuracy 欄位

# 5. 直接找出整張表裡面的最大值
max_value = acc_df.values.max()
print(f"——> 所有 accuracy 中的最大值為：{max_value:.6f}")

# 6. 如果還想知道這個最大值出現在第幾個 epoch、哪一個欄位
#    先把整個 acc_df 攤平成 1 維，再找 index
max_pos_flat = acc_df.values.argmax()

#    計算它對應到「第幾列（row）」與「第幾行（column）」
n_cols = acc_df.shape[1]
row_idx = max_pos_flat // n_cols   # 整除得到哪一“列”（對應 epochs 中的索引）
col_idx = max_pos_flat % n_cols    # 餘數得到哪一“行”（對應 acc_df.columns 中的索引）

max_epoch = epochs.iloc[row_idx]
max_column = acc_df.columns[col_idx]
print(f"——> 該最大值出現在 epoch={max_epoch}，欄位='{max_column}'。")

# 7. 如果想要每個欄位分別的最大值，可以這樣做：
max_per_column = acc_df.max(axis=0)  # 針對每一欄找最大值
print("——> 各個 Dataset 的最高 accuracy：")
print(max_per_column)

# 8. 若想要同時知道每個欄位最高值是哪一個 epoch：
idx_max_per_column = acc_df.idxmax(axis=0)   # 針對每一欄找出最大值所在的索引（這個索引對應到原本 df 的 row index）
epoch_max_per_column = epochs.iloc[idx_max_per_column.values].values
for col_name, col_max, col_rowidx in zip(
        acc_df.columns, 
        max_per_column.values, 
        idx_max_per_column.values
    ):
    print(f"    • 欄位 '{col_name}' 的最高 accuracy={col_max*100:.2f}，出現在 epoch={epochs.iloc[col_rowidx]}。")
```
